```python
import numpy as np
from collections import Counter

def calcular_estadisticas_airbnb(columna):
    # Paso 1: Limpiar los datos (quitar valores nulos / NaN)
    # Es vital porque si hay un nulo, la suma o el ordenamiento fallarán
    datos = [x for x in columna if x == x and x is not None]
    
    if not datos:
        return None, None, None

    # --- Media ---
    media = sum(datos) / len(datos)
    
    # --- Mediana ---
    datos_ordenados = sorted(datos)
    n = len(datos_ordenados)
    if n % 2 == 0:
        mediana = (datos_ordenados[n//2 - 1] + datos_ordenados[n//2]) / 2
    else:
        mediana = datos_ordenados[n//2]
    
    # --- Moda ---
    conteo = Counter(datos)
    max_freq = max(conteo.values())
    # Sacamos el primer valor de la lista de modas para que sea más fácil de leer
    moda = [k for k, v in conteo.items() if v == max_freq][0]
    
    return media, mediana, moda
```
